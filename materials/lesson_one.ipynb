## **1. 機械学習ってなに？どんなことができるの？**
**機械学習**とは、コンピュータが人間のように経験から学び、パターンを見つけて新しいデータに対応できるようになる仕組みです。人間が全てのルールをプログラムで書かなくても、大量のデータを使って「正解」を覚えさせることで、コンピュータ自身が判断できるようになります。
例えば、猫と犬の画像をたくさん見せて、それぞれのラベル（猫・犬）を教えると、新しい画像を見せたときに「これは猫っぽい」と判断できるようになります。また、天気や気温のデータから「明日の気温は何度になりそうか？」といった予測も可能になります。これが、私たちの生活の中でもたくさん活用されています。

📌 例：
- 写真を見せて「これは猫」「これは犬」と教えると、見分けられるようになる。
- 天気のデータを使って、明日の気温を予測する。

## **2. 人とコンピュータの違い、なぜ学習が必要？**
人間は経験や感覚を通して、あいまいなことも柔軟に判断できます。たとえば、「なんとなくこっちが正解っぽい」と感じたり、「ちょっと見たことあるパターンだからこうだろう」と推測したりします。
一方で、コンピュータはそういった直感的な判断ができません。はっきりとしたルールや数字がないと、正しく動くのが難しいです。そのため、コンピュータにとって「学習」はとても重要です。データをたくさん見て、そこから自動的にルールを見つけることで、初めて未知のデータにも対応できるようになります。（悪いデータを学習させると、パソコンは悪い結果を出すことになるので、いいデータを学習させた方がいいです。）

下記の表は上の話を表としてまとめたものです。

| **項目**           | **人間**                        | **コンピュータ**             |
|----------------|-----------------------------|--------------------------|
| 学び方         | 経験や感覚で学ぶ             | データと計算で学ぶ       |
| 曖昧さへの対応 | ある程度OK                  | 苦手（はっきりしてないと困る） |
| 得意なこと     | 直感、創造力、柔軟性         | 計算、記憶、正確さ         |

だからこそ、コンピュータに学ばせる方法が必要です！

## **3. 機械学習の3つのタイプ**

### **3.1. 分類（分類とは何か、日常の例）**
**分類**とは、ものごとをグループに分けることです。機械学習では、パソコンが「これはAのグループ」「これはBのグループ」「これはCのグループ」と判断できるように、たくさんのデータを使って学習させます。
例えば、Gmailで届いているメールをたくさん見せて、それぞれのラベル（正しい・迷惑メール）を教えると、新しいメールが来た時に「これは迷惑メールかも」と判断できるようになります。

- **分類**とは、物事を「グループ」に分ける学習
- 例：犬と猫の画像を見分ける、迷惑メールかどうか

### **3.2. 回帰（予測するとはどういうことか）**
**回帰**とは、数値（連続する値）を予測することです。例えば、明日の気温を予測する際に「明日は27.4度になりそう」という予測結果のようなものを**回帰**といいます。

- **回帰**とは、数字などの「数値（連続した値）」を予測する学習
- 例：明日の気温、家賃、売上の予測

また、分類と回帰の違いとしては下記のようになります。
| **分類** | **回帰** |
| :---: | :---: |
| 「これはA？B？」というグループ分け | 「いくつになる？」という数の予測 |
| 例：犬か猫か判断 | 例：家の家賃を予測 |


### **3.3. 強化学習（ゲームやロボットと学習）**
**強化学習**とは、「行動してみて、うまくいったらごほうびをもらう」というように、試行錯誤を通じて学んでいく方法です。人間がゲームで遊ぶときに「この動きをしたらうまくいった」「このやり方は失敗だった」と体験から覚えていくのと同じように、コンピュータも成功・失敗をもとに学びます。
例えば、迷路の中をゴールまで進むロボットに、最初はいろんな方向に進ませてみて、「ゴールに近づいたら +1点」「ぶつかったら -1点」というふうにルールを決めて、ごほうび（報酬）を与えながら学ばせます。

- **強化学習**とは、行動の結果から「良かった・悪かった」を学ぶ学習
- 例：迷路を解くゲーム、自動運転、ロボットの動き

## **4. 実例紹介（画像処理、音声認識、推薦システムなど）**
機械学習は、私たちの身の回りのさまざまな場所で使われています。ここでは、日常生活でよく見かける3つの例を紹介します。
**画像認識** では、写真や映像の中から「これは顔だ」「これは花だ」というように、物の種類を判別します。スマートフォンの顔認証や、カメラアプリが「これはチューリップ」「これはバラ」と花を判定してくれる機能も、画像認識の一種です。
**音声認識** は、人間の声を聞き取って、その内容を文字に変える技術です。たとえばスマートスピーカーに「今日の天気は？」と話しかけると、音声を文字に変換して理解し、天気を答えてくれます。
**推薦システム** は、あなたが過去に見たり買ったりしたものをもとに、「あなたに合いそうな商品や動画」を提案してくれる仕組みです。Amazonの「おすすめ商品」や、YouTubeの「あなたへのおすすめ」などがその例です。

- 📸 画像認識：カメラで見た物を見分ける（例：顔や花）
- 🎤 音声認識：話した言葉を聞き取って文字にする（例：スマートスピーカー）
- 🎯 推薦システム：好みに合う物を自動でおすすめしてくれる（例：Amazonのおすすめ）

## **🎲 アクティビティ**
### ① 絵カードゲーム：「これは何の動物？」
このゲームでは、いろいろな動物の絵カードを使って、分類のしくみを体験します。
最初に先生がカードを見せて、「これは何の動物かな？」と質問します。生徒は「猫！」「犬！」「うさぎ！」などと答えます。はじめは正解を教えながら進めていきますが、何度も繰り返すうちに、生徒はだんだん自分で考えて予測できるようになります。
これは、機械学習でいう「学習」と「予測」の流れと同じです。カードを何度も見て正解を覚え、次に見たカードが何かを当てる。つまり、「これはこのグループに入る！」という分類の体験をしているのです。
- **目的**: 動物の分類を通して、学習と予測の流れを体感する
- **流れ**: 先生がカードを見せる → 生徒が動物名を予測 → 答え合わせ → 繰り返して上達

## **💻 プログラミング：画像分類デモ（MobileNetV2）**
```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import decode_predictions, preprocess_input
from tensorflow.keras.preprocessing import image
import numpy as np
import matplotlib.pyplot as plt

# 画像の読み込み
img_path = "sample.jpg"  # サンプル画像を用意しておく
img = image.load_img(img_path, target_size=(224, 224))
x = image.img_to_array(img)
x = np.expand_dims(x, axis=0)
x = preprocess_input(x)

# モデルの読み込みと予測
model = MobileNetV2(weights="imagenet")
preds = model.predict(x)

# 結果表示
plt.imshow(img)
plt.axis('off')
plt.title("Prediction: " + decode_predictions(preds, top=1)[0][0][1])
plt.show()
```
