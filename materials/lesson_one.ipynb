## **1. 機械学習ってなに？どんなことができるの？**
**機械学習**とは、コンピュータが人間のように経験から学び、パターンを見つけて新しいデータに対応できるようになる仕組みです。人間が全てのルールをプログラムで書かなくても、大量のデータを使って「正解」を覚えさせることで、コンピュータ自身が判断できるようになります。
例えば、猫と犬の画像をたくさん見せて、それぞれのラベル（猫・犬）を教えると、新しい画像を見せたときに「これは猫っぽい」と判断できるようになります。また、天気や気温のデータから「明日の気温は何度になりそうか？」といった予測も可能になります。これが、私たちの生活の中でもたくさん活用されています。

📌 例：
- 写真を見せて「これは猫」「これは犬」と教えると、見分けられるようになる。
- 天気のデータを使って、明日の気温を予測する。

## **2. 人とコンピュータの違い、なぜ学習が必要？**
人間は経験や感覚を通して、あいまいなことも柔軟に判断できます。たとえば、「なんとなくこっちが正解っぽい」と感じたり、「ちょっと見たことあるパターンだからこうだろう」と推測したりします。
一方で、コンピュータはそういった直感的な判断ができません。はっきりとしたルールや数字がないと、正しく動くのが難しいです。そのため、コンピュータにとって「学習」はとても重要です。データをたくさん見て、そこから自動的にルールを見つけることで、初めて未知のデータにも対応できるようになります。（悪いデータを学習させると、パソコンは悪い結果を出すことになるので、いいデータを学習させた方がいいです。）

下記の表は上の話を表としてまとめたものです。

| 項目           | 人間                        | コンピュータ             |
|----------------|-----------------------------|--------------------------|
| 学び方         | 経験や感覚で学ぶ             | データと計算で学ぶ       |
| 曖昧さへの対応 | ある程度OK                  | 苦手（はっきりしてないと困る） |
| 得意なこと     | 直感、創造力、柔軟性         | 計算、記憶、正確さ         |

だからこそ、コンピュータに学ばせる方法が必要です！

## **3. 機械学習の3つのタイプ**

### **3.1. 分類（分類とは何か、日常の例）**
**分類**とは、ものごとをグループに分けることです。機械学習では、パソコンが「これはAのグループ」「これはBのグループ」「これはCのグループ」と判断できるように、たくさんのデータを使って学習させます。
例えば、Gmailで届いているメールをたくさん見せて、それぞれのラベル（正しい・迷惑メール）を教えると、新しいメールが来た時に「これは迷惑メールかも」と判断できるようになります。

- 物事を「グループ」に分ける学習
- 例：犬と猫の画像を見分ける、迷惑メールかどうか

### **3.2. 回帰（予測するとはどういうことか）**
- 数字などの「連続した値」を予測する学習
- 例：明日の気温、家賃、売上の予測

### **3.3. 強化学習（ゲームやロボットと学習）**
- 試行錯誤を通して「うまくなる」学習
- 例：迷路を解くゲーム、自動運転、ロボットの動き

## **💻 プログラミング：画像分類デモ（MobileNetV2）**
```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import decode_predictions, preprocess_input
from tensorflow.keras.preprocessing import image
import numpy as np
import matplotlib.pyplot as plt

# 画像の読み込み
img_path = "sample.jpg"  # サンプル画像を用意しておく
img = image.load_img(img_path, target_size=(224, 224))
x = image.img_to_array(img)
x = np.expand_dims(x, axis=0)
x = preprocess_input(x)

# モデルの読み込みと予測
model = MobileNetV2(weights="imagenet")
preds = model.predict(x)

# 結果表示
plt.imshow(img)
plt.axis('off')
plt.title("Prediction: " + decode_predictions(preds, top=1)[0][0][1])
plt.show()
```
